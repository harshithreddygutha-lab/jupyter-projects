```python
num = input("Enter a 4 digit number: ")

if len(num) == 4 and num == num[: :-1]:
    print("Palindrome")
else:
    print("Not a palindrome")
```

```python
num = int(input("Enter a 4 digit number: "))

if num < 1000 or num > 9999:
    print("Enter a 4 digit number")
else:
    original = num

    d1 = num % 10
    num = num //10

    d2 = num % 10
    num = num //10

    d3 = num % 10
    num = num//10

    d4 = num  % 10

    reverse = d1 * 1000 + d2 *100 + d3 * 10 + d4

    if original == reverse:
        print("Palindrome")
    else:
        print("Not a palindrome")
```
