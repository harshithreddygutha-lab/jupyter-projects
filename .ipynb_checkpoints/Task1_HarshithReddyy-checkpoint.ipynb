```python
n = 10
set_bit = n | (1<<1)
clear_bit = n & ~(1 << 0)
toggle_bit = n ^ (1<<2)

print("Original number:", n)
print("After setting 2nd bit:", set_bit)
print("After clearing 1st bit:", clear_bit)
print("After toggling 3rd bit:", toggle_bit)
```

```python
a = 10
b = 3

print("a & b =", a & b)
print("a | b =", a | b)
print("a ^ b =", a^ b)
print("~a =", ~a)
print("a << 2 =", a << 2)
print("a >> 1 =", a >> 1)
```

```python
def divisble_by_3_or_5(numbers):
    return[x for x in numbers if x % 3 ==0 or x % 5 ==0]

numbers = [1, 3, 5, 7, 9, 10, 12, 14, 15, 20]
print(divisble_by_3_or_5(numbers))
```

```python
x = 10

y = x >> 1
z = x << 2

print(y, z)
```

```python
Explanation: 10>>1 = 5
Right shift divides 1 and 2
10/2 = 5
10<<2 = 40
Left shift by 2 multiplies by 4
10*4 = 40
```
