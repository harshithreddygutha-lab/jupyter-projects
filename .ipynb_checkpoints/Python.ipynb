2.Version  
python 0.9.0(intial release (february 20 1991) [featuring classes,exception,handling,and core data types]   
python 1.0(released in january 1994) [added functional programming constructs lambda,map,filter]   
python 2.0(released in october 2000) [added list comphrensations,full unicode support,cycle detecting garbage]  
python 3.0(released in december 2010) [they fixed underlying design flaws]  
python 3.10(released in october 2011) [added structural pattern mapping]  
python 3.14(released in october 2005) [stable major series introducing (t strings),free threaded python]
